```python

import faiss, pickle
from sentence_transformers import SentenceTransformer
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Download sample lease
!wget -q -O sample_lease.pdf "https://eforms.com/images/2015/09/Residential-Lease-Agreement.pdf"

# Load + chunk
loader = PyPDFLoader("sample_lease.pdf")
documents = loader.load()
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""])
chunks = splitter.split_documents(documents)
print(f"Loaded {len(documents)} pages, split into {len(chunks)} chunks")

# Embed
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
texts = [c.page_content for c in chunks]
embeddings = model.encode(texts, show_progress_bar=True, convert_to_numpy=True)

# Build FAISS index
index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings.astype("float32"))
faiss.write_index(index, "faiss_index.index")
with open("chunk_texts.pkl", "wb") as f:
    pickle.dump(texts, f)
print(f"Index built with {index.ntotal} vectors")

# Sanity-test retrieval
def search(query, k=3):
    qvec = model.encode([query], convert_to_numpy=True).astype("float32")
    _, idxs = index.search(qvec, k)
    return [texts[i] for i in idxs[0]]

for r in search("what happens if I need to terminate under military orders?"):
    print(r[:300], "\n---")
```

```python

!pip install -q groq

from google.colab import userdata
from groq import Groq

client = Groq(api_key=userdata.get("GROK_API_KEY"))
```

```python

SYSTEM_PROMPT = """You are a document assistant. Answer the user's question strictly using the provided context below.
If the answer is not contained in the context, respond exactly with: "I cannot find this in the documents."
Do not use any outside knowledge or make assumptions beyond what is stated."""

def ask(query, k=3):
    context_chunks = search(query, k=k)          # reuse Phase 1's retrieval
    context = "\n\n".join(context_chunks)
    user_prompt = f"Context:\n{context}\n\nQuestion: {query}"

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": user_prompt}
        ],
        temperature=0.1
    )
    return response.choices[0].message.content

print(ask("what happens if I need to terminate under military orders?"))
print(ask("what is the penalty for late rent payment in nevada"))
```
